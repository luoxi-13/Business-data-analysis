```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import FuncFormatter

# ================= 1. 全局设置 =================
# 设置中文字体 (Windows用SimHei，Mac用Arial Unicode MS)
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False  # 正常显示负号

# ================= 2. 数据准备 =================
# 图1：年度销售趋势
years = [2017, 2018, 2019, 2020]
yearly_sales = [2931068, 3431930, 4243549, 5462455]

# 图2 & 图3：2020月度数据
months = list(range(1, 13))
sales_2020 = [222864, 285483, 399712, 333401, 632803, 565524, 340306, 588746, 502801, 577451, 468821, 544543]
growth_rate_monthly = [0.1995, 0.3336, 0.4801, 0.3398, 0.3693, 0.2169, 0.3414, 0.2471, 0.0784, 0.1605, -0.1235, 0.1502]

# 图4：11月各省份增长率
provinces = ['安徽', '北京', '福建', '甘肃', '广东', '广西', '贵州', '河北', '河南', '黑龙江', 
             '湖北', '湖南', '吉林', '江苏', '江西', '辽宁', '内蒙古', '宁夏', '青海', '山东', 
             '山西', '陕西', '上海', '四川', '天津', '西藏自治区', '新疆', '云南', '浙江']
growth_rate_nov = [0.1111, -0.8619, -0.0123, 0.9914, -0.2950, 0.7406, 1.0000, 0.8356, -1.1537, -0.6496, 
                   0.6024, 0.0515, -7.8032, -1.3197, 0.8362, 0.4518, -4.4243, 0.0000, 0.0000, -0.4176, 
                   0.6166, -1.4073, -2.8968, -8.2169, -0.4265, 1.0000, 1.0000, 0.4105, 0.4525]

# 图5：四川省月度增长率
sc_growth_rate = [0.6071, 0.5890, -0.5195, 0.4113, -0.2419, -1.3139, -3.4038, 0.7375, -6.7276, 0.4092, -8.2169, 0.9686]


# ================= 3. 绘图 =================
plt.figure(figsize=(12, 16)) 

# --- 图1：年度销售趋势分析 ---
plt.subplot(3, 2, 1)
plt.plot(years, yearly_sales, color='#5B9BD5', linewidth=2)
plt.title('年度销售趋势分析', fontsize=14, pad=10)
plt.ylim(0, 6500000)
plt.yticks([0, 2000000, 4000000, 6000000])
plt.xticks(years)
# 【关键修改】强制Y轴显示完整数字，不省略0
plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, pos: f'{int(x)}'))
plt.grid(axis='y', linestyle='-', alpha=0.3)

# --- 图2：2020年月销售趋势分析 ---
plt.subplot(3, 2, 2)
plt.plot(months, sales_2020, color='#5B9BD5', linewidth=2)
plt.title('2020年月销售趋势分析', fontsize=14, pad=10)
plt.ylim(0, 850000)
plt.yticks([0, 200000, 400000, 600000, 800000])
plt.xticks(months)

plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, pos: f'{int(x)}'))
plt.grid(axis='y', linestyle='-', alpha=0.3)

# --- 图3：2020年销售同比增长率趋势分析 ---
plt.subplot(3, 2, (3, 4))
plt.plot(months, growth_rate_monthly, color='#5B9BD5', linewidth=2)
plt.title('2020年销售同比增长率趋势分析', fontsize=14, pad=10)
plt.ylim(-0.2, 0.7)
plt.yticks([-0.2, 0.0, 0.2, 0.4, 0.6])
plt.xticks(months)
plt.grid(axis='y', linestyle='-', alpha=0.3)
plt.axhline(0, color='black', linewidth=0.5)

# --- 图4：2020年11月各省份同比增长情况分析 ---
plt.subplot(3, 2, (5, 6))
x_pos = np.arange(len(provinces))
plt.bar(x_pos, growth_rate_nov, color='#5B9BD5', width=0.6)
plt.title('2020年11月各省份同比增长情况分析', fontsize=14, pad=10)
plt.xticks(x_pos, provinces, rotation=45, ha='right', fontsize=9)
plt.ylim(-10, 2.5) 
yticks_val = [-10, -8, -6, -4, -2, 0, 2] 
plt.yticks(yticks_val, [f"{v*100:.2f}%" for v in yticks_val])
plt.grid(axis='y', linestyle='-', alpha=0.3)
plt.axhline(0, color='black', linewidth=0.5)

plt.tight_layout(pad=3.0)
plt.show()

# --- 图5：四川省2020年各月同比增长情况 ---
plt.figure(figsize=(12, 6))
plt.plot(months, sc_growth_rate, color='#5B9BD5', linewidth=2, marker='o', markersize=4)
plt.title('四川省2020年各月同比增长情况', fontsize=14, pad=10)
plt.ylim(-10, 2.5) 
yticks_val = [-10, -8, -6, -4, -2, 0, 2]
plt.yticks(yticks_val, [f"{v*100:.2f}%" for v in yticks_val])
plt.xticks(months)
plt.grid(axis='y', linestyle='-', alpha=0.5)
plt.axhline(0, color='black', linewidth=0.5)
plt.tight_layout()
plt.show()
```
