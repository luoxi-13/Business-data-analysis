```python
import matplotlib.pyplot as plt
import numpy as np

# ================= 1. 全局设置 =================
# 设置中文字体 (Windows用SimHei，Mac用Arial Unicode MS)
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False  # 正常显示负号

# ================= 2. 数据准备 =================
# 图1：购买客户的性别分布
gender_labels = ['Female', 'Male']
gender_data = [2371, 1555]
gender_colors = ['#5B9BD5', '#ED7D31'] # 蓝色和橙色

# 图2：女性消费客户的年龄段分布
age_labels = ['18-25岁', '26-35岁', '36-49岁', '50岁以上']
age_data = [265, 313, 412, 565]
age_colors = ['#5B9BD5', '#ED7D31', '#A5A5A5', '#FFC000'] # 蓝, 橙, 灰, 黄

# 图3：50岁以上的女性客户的消费品类分布
category_labels = ['Books', 'Clothing', 'Cosmetics', 'Food & Beverage', 'Shoes', 'Souvenir', 'Technology', 'Toys']
category_data = [57, 310, 135, 144, 99, 35, 42, 96]

# 图4：50岁以上女性服饰和护肤品消费人群的支付方式分布
# 注意：服饰(Clothing 310) + 护肤品(Cosmetics 135) = 445，与总计吻合
payment_labels = ['Alipay', 'Card', 'WeChat Pay']
payment_data = [150, 201, 94]

# ================= 3. 绘图 =================
# 创建画布，设置大小以匹配长图布局
fig = plt.figure(figsize=(12, 14))
fig.suptitle('某企业客群分析', fontsize=20, fontweight='bold', y=0.95)

color_bar = '#5B9BD5' # 统一柱状图颜色

# --- 图1：购买客户的性别分布 (左上) ---
ax1 = plt.subplot(3, 2, 1)
# 稍微分离饼图，模仿截图效果
explode1 = (0.05, 0.05) 
wedges, texts, autotexts = ax1.pie(gender_data, labels=None, autopct='%1.0f%%', 
                                   colors=gender_colors, explode=explode1, startangle=90)
ax1.set_title('购买客户的性别分布', fontsize=14, pad=10)
# 添加图例
ax1.legend(gender_labels, loc='center right', bbox_to_anchor=(1.2, 0.5))
# 添加边框
for spine in ax1.spines.values():
    spine.set_edgecolor('black')
    spine.set_linewidth(1.5)

# --- 图2：女性消费客户的年龄段分布 (右上) ---
ax2 = plt.subplot(3, 2, 2)
explode2 = (0.05, 0.05, 0.05, 0.05)
wedges, texts, autotexts = ax2.pie(age_data, labels=None, autopct='%1.0f%%', 
                                   colors=age_colors, explode=explode2, startangle=90)
ax2.set_title('女性消费客户的年龄段分布', fontsize=14, pad=10)
# 添加图例
ax2.legend(age_labels, loc='center right', bbox_to_anchor=(1.3, 0.5))
for spine in ax2.spines.values():
    spine.set_edgecolor('black')
    spine.set_linewidth(1.5)

# --- 图3：50岁以上的女性客户的消费品类分布 (中间) ---
ax3 = plt.subplot(3, 2, (3, 4))
x_pos = np.arange(len(category_labels))
ax3.bar(x_pos, category_data, color=color_bar, width=0.6)
ax3.set_title('50岁以上的女性客户的消费品类分布', fontsize=14, pad=10)
ax3.set_xticks(x_pos)
ax3.set_xticklabels(category_labels, rotation=45, ha='right') # 标签倾斜
ax3.set_ylim(0, 400)
ax3.set_yticks([0, 100, 200, 300, 400])
for spine in ax3.spines.values():
    spine.set_edgecolor('black')
    spine.set_linewidth(1.5)
# 添加网格线 (模仿截图)
ax3.grid(axis='y', linestyle='-', alpha=0.3)
ax3.set_axisbelow(True) # 网格线在柱子下面

# --- 图4：50岁以上女性服饰和护肤品消费人群的支付方式分布 (下方) ---
ax4 = plt.subplot(3, 2, (5, 6))
x_pos2 = np.arange(len(payment_labels))
ax4.bar(x_pos2, payment_data, color=color_bar, width=0.6)
ax4.set_title('50岁以上女性服饰和护肤品消费人群的支付方式分布', fontsize=14, pad=10)
ax4.set_xticks(x_pos2)
ax4.set_xticklabels(payment_labels)
ax4.set_ylim(0, 250)
ax4.set_yticks([0, 50, 100, 150, 200, 250])
for spine in ax4.spines.values():
    spine.set_edgecolor('black')
    spine.set_linewidth(1.5)
ax4.grid(axis='y', linestyle='-', alpha=0.3)
ax4.set_axisbelow(True)

plt.tight_layout(rect=[0, 0, 1, 0.93]) # 调整布局给大标题留空间
plt.show()
```
